## Setting: Oscar walks into a retail shop in Kampala

**The customer:** Sarah, owns a hardware store. 2 employees. Uses a notebook for sales. Internet is unreliable.

```python
# === SALES SCRIPT ===

opening = {
    "you": "Hi there! I'm Oscar. I noticed you're busy — how are you tracking your sales?",
    "goal": "Get them talking about their current pain point"
}

print(f"You: {opening['you']}")
print(f"Goal: {opening['goal']}")
```

---

## Objection 1: "How much does it cost?"

```python
def handle_pricing_objection():
    return {
        "customer": "How much does this thing cost?",
        "you": "It's completely free right now. We're launching and want businesses like yours to help shape the product. You'll never pay for what you use today.",
        "why_it_works": "Removes the price objection before it becomes a blocker. 'Free' is the strongest word in sales.",
        "follow_up": "Can I show you how it works offline?"
    }

result = handle_pricing_objection()
print(f"Customer: {result['customer']}")
print(f"You: {result['you']}")
print(f"Why: {result['why_it_works']}")
print(f"Follow up: {result['follow_up']}")
```

---

## Objection 2: "Does it work without internet?"

```python
def handle_offline_objection():
    return {
        "customer": "Internet here is very bad. If it goes down, can I still sell?",
        "you": "Yes — that's the main reason we built it. You process sales, print receipts, even check inventory — all without internet. When you're back online, everything syncs automatically. Nothing gets lost.",
        "why_it_works": "Offline is Custosell's strongest differentiator. Lead with it.",
        "follow_up": "Would you like to see how it works? I can show you right now."
    }

result = handle_offline_objection()
print(f"Customer: {result['customer']}")
print(f"You: {result['you']}")
print(f"Why: {result['why_it_works']}")
```

---

## Objection 3: "What about my staff?"

```python
def handle_staff_objection():
    return {
        "customer": "I have 2 employees. Can they use it too? What if they steal?",
        "you": "Each person gets their own login. You can see exactly who processed which sale, who closed the shift, and how much cash each person collected. If stock goes missing, you'll know.",
        "why_it_works": "Staff theft is a real fear for SME owners. Address it directly.",
        "follow_up": "I can set up different permissions — your manager sees reports, cashiers only see the sale screen."
    }

result = handle_staff_objection()
print(f"Customer: {result['customer']}")
print(f"You: {result['you']}")
```

---

## Objection 4: "Is my data safe?"

```python
def handle_data_objection():
    return {
        "customer": "Where is my data stored? Can anyone else see it?",
        "you": "Your data stays on your devices and our secure servers. We never share it. Not with anyone. Not for any reason. And you can export everything — sales, customers, inventory — anytime you want.",
        "why_it_works": "No jargon. No 'encryption' or 'compliance' talk. Simple promises they can understand.",
        "follow_up": "If you ever decide to leave Custosell, you take all your data with you. No lock-in."
    }

result = handle_data_objection()
print(f"Customer: {result['customer']}")
print(f"You: {result['you']}")
```

---

## Objection 5: "What if I want to switch later?"

```python
def handle_switching_objection():
    return {
        "customer": "What if I start using it and then want to switch to something else?",
        "you": "You can export all your data as Excel or CSV files anytime. Customers, sales history, inventory — everything. Custosell is built to let you leave if you want. But honestly, after a week of using it, you won't want to.",
        "why_it_works": "Confidence is contagious. Saying 'you can leave' actually makes them want to stay.",
        "follow_up": "Start with one month. If it doesn't work for you, export your data and go. No questions asked."
    }

result = handle_switching_objection()
print(f"Customer: {result['customer']}")
print(f"You: {result['you']}")
```

## The 3-Minute Pitch (if they ask you to just talk)

```python
pitch = """
Oscar, how do you track sales right now?
(pause — let them answer)

What happens when internet goes down?
(pause — let them answer)

Custosell works offline. You process sales, print receipts, 
track inventory — all without internet. When you're back online, 
everything syncs automatically.

Each staff member gets their own login. You know who sold what.
You see daily revenue, expenses, profit — on your phone or computer.

It's free to start. No contracts. No credit card. 
Your data is yours. You can leave anytime.

Can I show you how it works?
"""
print(pitch)
```

---

## Conversion Principles

| Principle | Why it works |
|-----------|-------------|
| **Start with their pain** | Don't pitch features. Ask how they do things now. Find the pain. |
| **Lead with offline** | It's Custosell's biggest advantage over every competitor. |
| **Free removes fear** | "It's free" is the easiest objection handler. |
| **Let them touch it** | A 2-minute demo is worth 30 minutes of talking. |
| **Data ownership = trust** | "You can export and leave anytime" builds more trust than any feature list. |
| **Close with a question** | Never end with "that's it." End with "Can I show you?" |
