## Standard Stand-Up

Before meaningful frontend work, Mike runs a short stand-up:

1. Mike defines success.
2. Sage identifies scope and reusable files.
3. Iris reviews UX and correction paths.
4. Blue proposes architecture.
5. Atlas stress-tests sync, persistence, routing, and auth risks.
6. Gauge defines diagnostics and error visibility.
7. Nora defines manual tests and regressions.
8. Rex confirms implementation files.
9. Vera defines verification gates.
10. Quill identifies documentation impact.

## Parallel Workflow

```text
Mike -> (Sage + Iris + Blue + Atlas + Gauge + Nora) -> Mike reconcile
     -> (Rex implementation lanes + Quill draft) -> Vera
     -> Rex fixes -> Quill final -> Mike final gate -> Oscar
```

## Mandatory Failure-State Review

Every offline or user-facing frontend flow must answer:

- What happens on validation failure?
- How can the user correct bad data?
- What happens on retry, duplicate submit, stale cache, and failed sync?
- Is the error message actionable?
- Does the flow preserve offline-first behavior?
